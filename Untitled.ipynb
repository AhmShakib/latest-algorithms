```python
import matplotlib.pyplot as plt
import numpy as np

# আপনার মডেলের ফলাফলগুলো
# (এগুলো আপনি আপনার মডেল প্রশিক্ষণ চালানোর পর test_losses এবং test_maes থেকে পাবেন)
# উদাহরণস্বরূপ, GRU মডেলের বিভিন্ন WINDOW সাইজের ফলাফল:
window_sizes = ['Window=3', 'Window=6', 'Window=9', 'Window=12']
mse_values = [0.0066, 0.0076, 0.0100, 0.0099]
mae_values = [0.0646, 0.0724, 0.0868, 0.0852]

# কাস্টম কালার প্যালেট (আপনার আগের পছন্দ অনুযায়ী)
color_mse = '#0A2350' # গাঢ় নীল/নেভি ব্লু
color_mae = '#1E408A' # মধ্যম নীল

x = np.arange(len(window_sizes))  # x-axis এর লেবেল পজিশন
width = 0.35  # বারের প্রস্থ

fig, ax = plt.subplots(figsize=(10, 6))

# MSE বারের প্লট
rects1 = ax.bar(x - width/2, mse_values, width, label='Test MSE', color=color_mse)
# MAE বারের প্লট
rects2 = ax.bar(x + width/2, mae_values, width, label='Test MAE', color=color_mae)

# প্লটের টাইটেল, লেবেল এবং লেজেন্ড যোগ করা
ax.set_ylabel('Metric Value')
ax.set_title('Model Performance for Different Window Sizes (GRU Model)')
ax.set_xticks(x)
ax.set_xticklabels(window_sizes)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7) # y-অ্যাক্সিসে গ্রিড যোগ করা

# বারের উপরে মানগুলো যোগ করা (ঐচ্ছিক, কিন্তু পরিষ্কার দেখায়)
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.4f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')

autolabel(rects1)
autolabel(rects2)

plt.tight_layout() # লেবেল এবং টাইটেল যাতে প্লটের বাইরে না যায়
plt.show()
```
